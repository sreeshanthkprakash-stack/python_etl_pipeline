```python
df_load = df.astype(object).where(pd.notna(df), None)
```

```python
sql = """ INSERT INTO weather2 (
    observation_date,
    observation_time,
    wind_direction,
    wind_speed_kmh,
    wind_gust_speed_kmh,
    temperature_c,
    dew_point_c,
    feels_like_temperature_c,
    relative_humidity_percent,
    fire_index,
    rainfall_mm,
    rainfall_10min_mm,
    air_pressure_hpa
)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s);"""
```

```python
for _, row in df_load.iterrows():
    cursor.execute(sql, (
        row["observation_date"],
        row["observation_time"],
        row["wind_direction"],
        row["wind_speed"],
        row["wind_gust_speed"],
        row["temperature"],
        row["dew_point"],
        row["feels_like_temperature"],
        row["relative_humidity"],
        row["fire_index"],
        row["rainfall"],
        row["rainfall_10min"],
        row["air_pressure"]
    ))
```
